# SmartHire — Supervised Resume Category Classifier (Model A)

Trains and evaluates a multi-class text classification model predicting candidate role domains using TF-IDF and Logistic Regression / Random Forest.


```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.config import CLEAN_RESUMES_PATH, CLASSIFIER_PATH
from src.models.classifier import train_classifier, predict_category, load_classifier

resumes = pd.read_csv(CLEAN_RESUMES_PATH)
print(f"Loaded {len(resumes)} labeled resumes.")
```

### Train & Evaluate Classifier


```python
model, metrics = train_classifier(resumes, save=True)
print(f"Test Accuracy: {metrics['accuracy']:.4f}")
print(f"Weighted F1-Score: {metrics['f1_weighted']:.4f}")
print(f"Macro F1-Score: {metrics['f1_macro']:.4f}")
```

### Sample Inference Test


```python
sample_cv = "Senior Software Engineer with 6 years experience in Python, Django, React, PostgreSQL, Docker, AWS and microservices architecture."
result = predict_category(model, sample_cv)
print("Predicted Domain:", result['predicted_category'])
print(f"Confidence: {result['confidence_pct']}%")
print("Top 3 Predictions:", result['top_predictions'][:3])
```
